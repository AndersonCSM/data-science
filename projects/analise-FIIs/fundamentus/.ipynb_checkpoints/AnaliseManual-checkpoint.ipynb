# Criando um algoritmo para analisar todas as planilhas e criar uma tabela de frequência

```python
# Criando a função tabela
def tabela(frequencia, dt):
    for nome in dt.papel:
        if nome not in frequencia:
            frequencia[nome] = 1
        else:
            frequencia[nome] += 1

    return frequencia
```

```python
# caminho da pasta
path = 'planilhas_fundamentus_fii/'
```

```python
# Percorrendo cada arquivo da pasta e criando a tabela de frequência

frequencia = {}

qtd_arquivos = 0 

for nome_arquivo in os.listdir(path):
    
    # Verificando se o arquivo é um arquivo csv
    if nome_arquivo.endswith(".csv"):
        
        # Concatenando o caminho para o arquivo para repassar ao Pandas
        arquivo = os.path.join(path, nome_arquivo)
        
        # Transformando em um Dataframe
        dt = pd.read_csv(arquivo)
        
        # Removendo fundos com valores inconstantes
        dt.replace([np.inf, -np.inf], np.nan, inplace=True)
        dt.dropna(inplace=True)
        
        # Aplicando filtro para selecionar os fundos que apresentam ffo_yield > 0, dividend_yield > 0, p_sobre_vp > 0, valor_de_mercado > 0 e liquidez > 0
        dt = dt[((dt.ffo_yield > 0) & (dt.dividend_yield > 0) & (dt.p_sobre_vp > 0) & (dt.valor_de_mercado > 0) & (dt.liquidez > 0))]

        
        # aplicando o filtro que o preço da cota deve ser menor que R$ 125,00 pois é o que eu disponho para comprar atualmente
        dt = dt[dt.cotacao < 125]
        
        # Aplicando filtro que o rendimento do fundo é maior do que ele provém de cota, permitindo estabilidade nos pagamentos
        dt = dt[dt.ffo_yield >= dt.dividend_yield]
        
        # Aplicando o filtro de que o p/vp seja menor do que 1, logo estária bom para comprar
        dt = dt[dt.p_sobre_vp <= 1]

        # Ordenando por valor crescente da quantidade de cotas
        dt = dt.sort_values(by='cotacao', ascending=True)

        frequencia = tabela(frequencia, dt)
        
        qtd_arquivos += 1
```

```python
# descarto KNRE11 por pagar dividendos a cada 6 meses

frequencia
```

```python
x = list(frequencia.keys())
y = list(frequencia.values())
dados = [x, y]
```

```python
dt_freq = pd.DataFrame(data = dados)
```

```python
dt_freq = dt_freq.transpose()
```

```python
dt_freq.columns = ["nome","contagem"]
```

```python
dt_freq
```

```python
dt_freq = dt_freq[dt_freq.contagem >= qtd_arquivos]
dt_freq
```

```python
dt_recent = dt_recent[dt_recent.papel.isin(dt_freq.nome)]
```
